```python
import pandas as pd
import subprocess
```

```python
col_titles = ['Run','AvgSpotLen','Bases','BioProject','BioSample','Experiment','Instrument','Library Name','LibraryLayout','LibrarySelection','LibrarySource','Organism','Platform','SRA Study', 'FileLocations']
df = pd.DataFrame(columns=col_titles)
```

```python
# puts all of the sample folder names from PM24 into a list and gets rid of any standalone files and any extra whitespaces newlines
# Modify the dropbox path when creating a RunInfoFile for a new run
sample_files = subprocess.check_output(['rclone', 'lsf', 'ptm_dropbox:/CoS/BioSci/BioSci-McGrath/Apps/CichlidSequencingData/Reads/PM27', '--dirs-only'], encoding = 'utf-8').strip().split('\n')
sample_files
```

```python
# Use this to verify that the number of dirs found is how many there should be!!
len(sample_files)
```

```python
run = []
concise_samples = []
for sample in sample_files:
    concise_samples.append(sample.split('.')[0][:-3])
len(run)
df['Run'] = concise_samples
concise_samples
```

```python
# defining lots of the same variables here.
# Change these as needed when maiking RunInfoFiles for new projects. 
AvgSpotLen = '300'
Bases= '0'
BioProject = 'low_coverage_fish'
Experiment = 'low_coverage_fish'
Instrument = 'Element AVITI'
LibraryLayout = 'PAIRED'
LibrarySelection = '-'
LibrarySource = 'GENOMIC'
Platform = 'ELEMENT Cloudbreak FS' 
SRA_Study = 'MappingYHSexDeterminer'
```

```python
df['AvgSpotLen'] = [AvgSpotLen]*len(df)
df['Bases'] = [Bases]*len(df)
df['BioProject'] = [BioProject]*len(df)
df['Experiment'] = [Experiment]*len(df)
df['Instrument'] = [Instrument]*len(df)
df['LibraryLayout'] = [LibraryLayout]*len(df)
df['LibrarySelection'] = [LibrarySelection]*len(df)
df['LibrarySource'] = [LibrarySource]*len(df)
df['Platform'] = [Platform]*len(df)
df['SRA Study'] = [SRA_Study]*len(df)
```

```python
# to remove 'lab_' prefixes and '_L001' and '_L002' suffixes
df['BioSample'] = concise_samples # from above
df['BioSample'] = df.BioSample.replace('lab_', '', regex=True)
df['BioSample'] = df.BioSample.replace('_L001', '', regex=True)
df['BioSample'] = df.BioSample.replace('_L002', '', regex=True)
```

```python
# add the 'Library Name' Column

Library_Name = df['BioSample'] + '_library'
df['Library Name'] = Library_Name
# df
```

```python
# Add the FileLocations
# also removes any non fastq files from the paths and prints length to confirm the number of paths you have
# Be sure to update the rclone path with the reads you're making the RunInfoFile for 
paths = subprocess.check_output(['rclone', 'lsf', 'ptm_dropbox:/CoS/BioSci/BioSci-McGrath/Apps/CichlidSequencingData/Reads/PM27', '--format', 'p', '--files-only', '-R'], encoding = 'utf-8').strip().split('\n')
paths = [x for x in paths if x.endswith('.fastq.gz')]
print('Number of FASTQ files is: ' + str(len(paths)))
```

```python
# Be sure to update the start directory in LocalReadsDir path 
paths
concat_paths = []
for path1,path2 in zip(*[iter(paths)]*2):
    sample_paths = []
    sample_paths.append('PM27/'+ path1)
    sample_paths.append('PM27/' + path2)
    joined_paths = ',,'.join(sample_paths)
    concat_paths.append(joined_paths)
print("Total Samples: " + str(len(concat_paths)))
concat_paths
```

```python
# Add Species based on BioSample Names
# This will need to get updated whenever more species are present than are available below
species_map = {
    "YH": "Aulonocara sp. 'chitande type north' Nkhata Bay",
    "MC": "Mchenga conophoros",
    "MCYHF1": "Mchenga conophoros x Aulonocara sp. 'chitande type north' Nkhata Bay F1 hybrid",
    "MCYHBC1": "Mchenga conophoros x Aulonocara sp. 'chitande type north' Nkhata Bay BC1 hybrid",
}
```

```python
# for each sample, find its first species and assign it into the Organism column. 
df.set_index(df.Run, inplace=True) # set Run column as index 
for sample in concise_samples:
    hyphen_sample = sample.replace('_', '-')
    species_prefix = hyphen_sample.split('-')[0]
    # print(df.loc[sample]["Organism"])
    # print(species_map[species_prefix])
    df.loc[sample, "Organism"] = species_map[species_prefix]
```

```python
# Do the complicated, yet not so complicated filepath mapping magic
# Be sure to check each sample to make sure that this is correct before running the pipelines!!!!!!!!
for sample in concise_samples:
    for path in concat_paths:
        if sample in path:
            df.loc[df.Run == sample, 'FileLocations'] = path
```

```python
# send to excel!
# Double check that the "Organism" column has data for all rows
df.to_excel('/home/mcgrath-lab/nkumar317/CichlidSRSequencing/cichlid_sr_sequencing/analysis_pipeline/RunInfoFiles/PM27_LocalData.xlsx', index=False)
df.to_csv('/home/mcgrath-lab/nkumar317/CichlidSRSequencing/cichlid_sr_sequencing/analysis_pipeline/RunInfoFiles/PM27_LocalData.csv', index=False)
```
